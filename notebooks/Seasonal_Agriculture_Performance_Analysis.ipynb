### Importing Libraries

### Load Dataset

```python
# ============================================================
# LOAD DATASET
# ============================================================

file_path = "../data/seasonal_agriculture_performance_dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

### Statistical Summary

```python
# ============================================================
# STATISTICAL SUMMARY
# ============================================================

df.describe().T
```

### Missing Values

```python
# ============================================================
# MISSING VALUE ANALYSIS
# ============================================================

missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values,
    "Missing Percentage": (
        missing_values.values / len(df) * 100
    ).round(2)
})

missing_table = missing_table[
    missing_table["Missing Values"] > 0
]

missing_table
```

### Duplicate Check

```python
# ============================================================
# DUPLICATE CHECK
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

### Categorical Variables

```python
# ============================================================
# CATEGORICAL VARIABLE EXPLORATION
# ============================================================

categorical_columns = [
    "State",
    "Crop",
    "Season",
    "Irrigation_Method"
]

for column in categorical_columns:
    print("\n" + "=" * 60)
    print(column)
    print("=" * 60)
    print(df[column].value_counts())
```

### Create Clean Dataset

```python
# ============================================================
# CREATE CLEAN DATASET
# ============================================================

df_clean = df.copy()

print("Original dataset shape :", df.shape)
print("Working dataset shape  :", df_clean.shape)
```

### Handle Missing Values

```python
# ============================================================
# HANDLE MISSING NUMERICAL VALUES
# ============================================================

columns_to_impute = [
    "Rainfall_mm",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

for column in columns_to_impute:
    median_value = df_clean[column].median()
    df_clean[column] = df_clean[column].fillna(median_value)

print("Missing numerical values handled successfully.")
```

### Verify Cleaning

```python
# ============================================================
# VERIFY DATA CLEANING
# ============================================================

remaining_missing = df_clean.isnull().sum().sum()

print("Total missing values after cleaning:", remaining_missing)
```

### Check Duplicates After Cleaning

```python
# ============================================================
# FINAL DUPLICATE CHECK
# ============================================================

duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows after cleaning:", duplicate_count)
```

### Final Dataset Overview

```python
# ============================================================
# FINAL CLEAN DATASET OVERVIEW
# ============================================================

print("FINAL CLEANED DATASET")
print("=" * 60)

print("Rows              :", df_clean.shape[0])
print("Columns           :", df_clean.shape[1])
print("Missing Values    :", df_clean.isnull().sum().sum())
print("Duplicate Rows    :", df_clean.duplicated().sum())

print("\nSeasons:")
print(df_clean["Season"].unique())

print("\nCrops:")
print(df_clean["Crop"].unique())

print("\nNumber of States:")
print(df_clean["State"].nunique())

print("\nIrrigation Methods:")
print(df_clean["Irrigation_Method"].unique())
```

### Save Clean Dataset

```python
# ============================================================
# SAVE CLEANED DATASET
# ============================================================

output_path = "../outputs/tables/cleaned_seasonal_agriculture_dataset.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print(output_path)
```

## SEASONAL PERFORMANCE ANALYSIS

### Check Season Distribution

```python
# ============================================================
# SEASON-WISE RECORD DISTRIBUTION
# ============================================================

season_counts = df_clean["Season"].value_counts()

print("Number of records by season:")
print(season_counts)
```

### Create Seasonal Summary

```python
# ============================================================
# SEASONAL PERFORMANCE SUMMARY
# ============================================================

season_summary = df_clean.groupby("Season").agg(
    Average_Yield_Tonnes_Ha=("Yield_Tonnes_Ha", "mean"),
    Average_Profit_INR=("Profit_INR", "mean"),
    Average_Water_Efficiency=("Water_Used_m3", "mean"),
    Average_Disease_Pest_Risk=("Disease_Pest_Risk_pct", "mean"),
    Average_Rainfall_mm=("Rainfall_mm", "mean"),
    Average_Soil_Moisture_pct=("Soil_Moisture_pct", "mean")
).round(2)

season_summary
```

### Calculate ROI

```python
# ============================================================
# ROI CALCULATION
# ============================================================

df_clean["ROI_pct"] = (
    df_clean["Profit_INR"] /
    df_clean["Total_Cost_INR"]
) * 100

roi_summary = (
    df_clean.groupby("Season")["ROI_pct"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("Average ROI by Season")
print("=" * 40)
print(roi_summary)
```

### Complete Seasonal Summary

```python
# ============================================================
# COMPLETE SEASONAL PERFORMANCE SUMMARY
# ============================================================

season_summary["Average_ROI_pct"] = (
    df_clean.groupby("Season")["ROI_pct"].mean()
)

season_summary = season_summary.round(2)

season_summary
```

### Average Yield by Season

```python
# ============================================================
# CHART 1 - AVERAGE YIELD BY SEASON
# ============================================================

plt.figure(figsize=(9, 6))

ax = sns.barplot(
    data=df_clean,
    x="Season",
    y="Yield_Tonnes_Ha",
    estimator="mean",
    errorbar=None
)

plt.title(
    "Average Agricultural Yield by Season",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Season", fontsize=12)
plt.ylabel("Average Yield (Tonnes/Ha)", fontsize=12)

# Add values on bars
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.2f",
        padding=3
    )

plt.tight_layout()

plt.savefig(
    "../outputs/charts/06_season_yield.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CHART 17 - CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(13, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    linewidths=0.3,
    center=0
)

plt.title(
    "Correlation Between Agricultural Performance Variables",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    "../outputs/charts/22_correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Yield Correlations

```python
# ============================================================
# VARIABLES MOST RELATED TO YIELD
# ============================================================

yield_correlations = (
    correlation_matrix["Yield_Tonnes_Ha"]
    .drop("Yield_Tonnes_Ha")
    .sort_values(ascending=False)
)

print("CORRELATION WITH YIELD")
print("=" * 50)
print(yield_correlations.round(3))
```

### Profit Correlations

```python
# ============================================================
# VARIABLES MOST RELATED TO PROFIT
# ============================================================

profit_correlations = (
    correlation_matrix["Profit_INR"]
    .drop("Profit_INR")
    .sort_values(ascending=False)
)

print("CORRELATION WITH PROFIT")
print("=" * 50)
print(profit_correlations.round(3))
```

### Strongest Relationships

```python
# ============================================================
# STRONGEST CORRELATIONS
# ============================================================

corr_pairs = (
    correlation_matrix
    .where(
        np.triu(
            np.ones(correlation_matrix.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .sort_values(
        key=abs,
        ascending=False
    )
)

print("TOP 10 STRONGEST VARIABLE RELATIONSHIPS")
print("=" * 60)

print(corr_pairs.head(10).round(3))
```
